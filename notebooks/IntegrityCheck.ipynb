```python
# ── UNSW-NB15 Integrity Check ──────────────────────────────
import pandas as pd
import numpy as np
import os

UNSW_PATH = r"C:\Users\alifa\OneDrive\Desktop\dissertation_project\data\UNSW-NB15"

print("=" * 60)
print("UNSW-NB15 DATASET INTEGRITY CHECK")
print("=" * 60)

# ── Step 1: Check all expected files are present ───────────
expected_files = [
    'UNSW-NB15_1.csv',
    'UNSW-NB15_2.csv',
    'UNSW-NB15_3.csv',
    'UNSW-NB15_4.csv',
    'UNSW-NB15_features.csv',
    'UNSW-NB15_GT.csv'
]

print("\nStep 1 — File presence check:")
all_present = True
for f in expected_files:
    path = os.path.join(UNSW_PATH, f)
    if os.path.exists(path):
        size = os.path.getsize(path) / (1024*1024)
        print(f"  ✓ {f:<35} ({size:.1f} MB)")
    else:
        print(f"  ✗ {f:<35} MISSING!")
        all_present = False

if all_present:
    print("\n  All expected files present!")
else:
    print("\n  Some files are missing — please re-download them")

# ── Step 2: Load and check each data file ──────────────────
print("\nStep 2 — Loading and checking each CSV file:")
dfs = []
total_rows = 0

for f in ['UNSW-NB15_1.csv', 'UNSW-NB15_2.csv',
          'UNSW-NB15_3.csv', 'UNSW-NB15_4.csv']:
    path = os.path.join(UNSW_PATH, f)
    if os.path.exists(path):
        try:
            df_temp = pd.read_csv(path, header=None, low_memory=False)
            rows, cols = df_temp.shape
            total_rows += rows
            nulls = df_temp.isnull().sum().sum()
            print(f"  {f}")
            print(f"    Rows: {rows:,}  |  Columns: {cols}  |  Nulls: {nulls:,}")
            dfs.append(df_temp)
        except Exception as e:
            print(f"  ERROR loading {f}: {e}")

# ── Step 3: Load feature names ─────────────────────────────
print("\nStep 3 — Feature names file:")
feat_path = os.path.join(UNSW_PATH, 'UNSW-NB15_features.csv')
if os.path.exists(feat_path):
    features_df = pd.read_csv(feat_path)
    print(f"  Loaded successfully")
    print(f"  Number of features defined: {len(features_df)}")
    print(f"  Columns in feature file: {list(features_df.columns)}")

# ── Step 4: Combine and inspect ────────────────────────────
print("\nStep 4 — Combined dataset summary:")
if dfs:
    df_unsw = pd.concat(dfs, ignore_index=True)
    print(f"  Total rows    : {len(df_unsw):,}")
    print(f"  Total columns : {df_unsw.shape[1]}")

    # Check for duplicates
    dupes = df_unsw.duplicated().sum()
    print(f"  Duplicate rows: {dupes:,}")

    # Check for nulls
    nulls = df_unsw.isnull().sum().sum()
    print(f"  Total nulls   : {nulls:,}")

    # Check for infinite values
    numeric_cols = df_unsw.select_dtypes(include=[np.number]).columns
    infs = np.isinf(df_unsw[numeric_cols].values).sum()
    print(f"  Infinite values: {infs:,}")

    # Check last column (usually the label)
    last_col = df_unsw.iloc[:, -1]
    print(f"\n  Last column (label) value counts:")
    print(f"  {last_col.value_counts().to_dict()}")

# ── Step 5: Ground truth file ──────────────────────────────
print("\nStep 5 — Ground truth file:")
gt_path = os.path.join(UNSW_PATH, 'UNSW-NB15_GT.csv')
if os.path.exists(gt_path):
    gt_df = pd.read_csv(gt_path)
    print(f"  Loaded successfully")
    print(f"  Shape: {gt_df.shape}")
    print(f"  Columns: {list(gt_df.columns)}")
    print(f"  Attack categories:")
    for col in gt_df.columns:
        if gt_df[col].dtype == object:
            print(f"    {gt_df[col].value_counts().to_dict()}")

print("\n" + "=" * 60)
print("INTEGRITY CHECK COMPLETE")
print("=" * 60)
```

```python
# ── UNSW-NB15 Integrity Check — Continued ──────────────────
import pandas as pd
import numpy as np
import os

UNSW_PATH = r"C:\Users\alifa\OneDrive\Desktop\dissertation_project\data\UNSW-NB15"

# ── Step 3: Load feature names with correct encoding ───────
print("Step 3 — Feature names file:")
feat_path = os.path.join(UNSW_PATH, 'UNSW-NB15_features.csv')
if os.path.exists(feat_path):
    try:
        features_df = pd.read_csv(feat_path, encoding='latin-1')
        print(f"  Loaded successfully")
        print(f"  Number of features defined: {len(features_df)}")
        print(f"  Columns in feature file: {list(features_df.columns)}")
        print(f"\n  Feature list:")
        for _, row in features_df.iterrows():
            print(f"    {row.iloc[0]:3} | {str(row.iloc[1]):<35} | {str(row.iloc[2])}")
    except Exception as e:
        print(f"  ERROR: {e}")

# ── Step 4: Load and combine all data files ─────────────────
print("\nStep 4 — Combined dataset summary:")
dfs = []
for f in ['UNSW-NB15_1.csv', 'UNSW-NB15_2.csv',
          'UNSW-NB15_3.csv', 'UNSW-NB15_4.csv']:
    path = os.path.join(UNSW_PATH, f)
    if os.path.exists(path):
        df_temp = pd.read_csv(path, header=None,
                              low_memory=False,
                              encoding='latin-1')
        dfs.append(df_temp)

df_unsw = pd.concat(dfs, ignore_index=True)
print(f"  Total rows     : {len(df_unsw):,}")
print(f"  Total columns  : {df_unsw.shape[1]}")

# Duplicates
dupes = df_unsw.duplicated().sum()
print(f"  Duplicate rows : {dupes:,}")

# Nulls per column
null_counts = df_unsw.isnull().sum()
null_cols = null_counts[null_counts > 0]
print(f"  Columns with nulls: {len(null_cols)}")
if len(null_cols) > 0:
    for col, count in null_cols.items():
        print(f"    Column {col}: {count:,} nulls")

# Infinite values
numeric_cols = df_unsw.select_dtypes(include=[np.number]).columns
infs = np.isinf(df_unsw[numeric_cols].values).sum()
print(f"  Infinite values: {infs:,}")

# ── Step 5: Ground truth file ───────────────────────────────
print("\nStep 5 — Ground truth file:")
gt_path = os.path.join(UNSW_PATH, 'UNSW-NB15_GT.csv')
if os.path.exists(gt_path):
    try:
        gt_df = pd.read_csv(gt_path, encoding='latin-1')
        print(f"  Loaded successfully")
        print(f"  Shape: {gt_df.shape}")
        print(f"  Columns: {list(gt_df.columns)}")
        print(f"\n  Attack category distribution:")
        for col in gt_df.columns:
            if gt_df[col].dtype == object or gt_df[col].nunique() < 20:
                print(f"\n  {col}:")
                print(gt_df[col].value_counts().to_string())
    except Exception as e:
        print(f"  ERROR: {e}")

# ── Step 6: Inspect last column (label) ─────────────────────
print("\nStep 6 — Label column (last column) distribution:")
last_col = df_unsw.iloc[:, -1]
print(f"  Unique values: {last_col.nunique()}")
print(f"  Value counts:\n{last_col.value_counts().to_string()}")

print("\n" + "=" * 60)
print("INTEGRITY CHECK COMPLETE")
print("=" * 60)
```
